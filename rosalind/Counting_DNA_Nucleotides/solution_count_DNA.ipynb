# Решение задачи Counting DNA Nucleotides
Дано: последовательность ДНК в виде одной строчки файла .txt

Задача: посчитать число вхождений каждого нуклеотида в данную строчку

Решение:



```python
with open('rosalind_dna2.txt') as file:
    s = file.read()
count_a = s.count('A')
count_t = s.count('T')
count_c = s.count('C')
count_g = s.count('G')
print(count_a, count_c, count_g, count_t, sep = ' ')
```
