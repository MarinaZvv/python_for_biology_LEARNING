# Решение задачи Complementing a Strand of DNA

Дано: последовательность ДНК

Задача: получить обратный комплемент (т.е. комплементарную последовательность, которую потом нужнл развернуть задом наперед)

```python
with open('rosalind_revc.txt') as file:
    DNA_1 = file.read().strip()
```

```python
transctiption = {'A': 'T',
                 'T': 'A',
                 'C': 'G',
                 'G': 'C'}
```

```python
DNA_2 = ''
for nuc in DNA_1:
    DNA_2 += transctiption[nuc]
print(DNA_2[::-1])
```
