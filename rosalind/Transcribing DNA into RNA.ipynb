# Решение задачи Transcribing DNA into RNA
Дано: последовательность ДНК в виде строки (смысловая цепь)

Задача: из этой цепочки ДНК сделать цепочку РНК путем заметы тимина на урацил

Решение:

```python
with open('rosalind_rna.txt') as file:
    DNA = file.read()

RNA = DNA.replace('T', 'U') #работа метода: заменяет первый аргумент на второй
print(RNA)
```

```python
# без метода
with open('rosalind_rna.txt') as file:
    DNA = file.read()

RNA = ''
for nucl in DNA:
    if nucl != 'T':
        RNA += nucl
    else:
        nucl = 'U'
        RNA += nucl

print(RNA)
```

```python
# можно было чуть менее громоздко:
with open('rosalind_rna.txt') as file:
    DNA = file.read()

RNA = ''
for nucl in DNA:
    if nucl == 'T':
        RNA += 'U'
    else:
        RNA += nucl

print(RNA)
```
